##Null Hypothesis

The mean weekly cost of restaurants follows the theoretical cost model:
The actual mean cost matches the expected cost from the model.                                                                        
##Alternative Hypothesis

The mean weekly cost of restaurants is different from the theoretical cost model:

The actual mean cost significantly  higher the model.

###This is a two-tailed test since we are testing for any significant difference (higher or lower) from the theoretical mean cost.

```python
import scipy.stats as stats
import numpy as np
import pandas as pd
```

```python
# Given data
W = 4000  # Theoretical mean weekly cost
sample_mean = 3050  # Sample mean weekly cost
stdev = 125  # Standard deviation of weekly operating cost
n = 25  # Sample size
df = n - 1  # Degrees of freedom
```

```python
std_error = stdev / np.sqrt(n)
print('Standard error is',std_error)
stats.t.interval(0.95, df = 24)
```

```python
#95%of the times average cost must be between the range given below
print('Lower limit of 95% average weekly cost', round(sample_mean+(stats.t.interval(0.95, df = 24)[0])*(std_error),3))
print('Upper limit of 95% average weekly cost', round(sample_mean+(stats.t.interval(0.95, df = 24)[1])*(std_error),3))
```

```python
#Right hand side probability
round(1-stats.t.cdf(stats.t.interval(0.95, df = 24)[1], df=24),3)
```

```python
# Left hand side probability
stats.t.cdf(stats.t.interval(0.95, df = 24)[1], df=79)
```

```python
# Calculation of Test Statistic
t=(3050-4000)/(stdev/np.sqrt(n))
print('Value of Test Statistic is',t)

#The probability given as output is for the average cost of <=3050
print('Probability of the average weekly cost being <= 3050 is',(stats.t.cdf(t, df=24)))


# To calculate the pvalue of the average cost being >= 3050 is
print('Pvalue >= 3050 is',round(1-stats.t.cdf(t, df=24),2))

# Determine the critical values for a two-tailed test at alpha level 5% (alpha = 0.05)
critical_value_lower = stats.t.ppf(0.025, df)
critical_value_upper = stats.t.ppf(0.975, df)
print('Critical Values at = 0.05 are', critical_value_lower, 'and', critical_value_upper)

# Make a decision
if t < critical_value_lower or t > critical_value_upper:
    print('Reject the null hypothesis.')
else:
    print('Fail to reject the null hypothesis.')


# Conclusion
if t < critical_value_lower or t > critical_value_upper:
    print('There is strong evidence to support the restaurant owners claim that the weekly operating costs are different from the model suggests.')
else:
    print('There is not enough evidence to support the restaurant owners claim that the weekly operating costs are different from the model suggests.')

```

##Decision: Given the very low probability of obtaining the sample mean of 3050 or lower if the true mean were 4000, we would reject the null hypothesis. There is strong evidence to suggest that the true mean weekly cost is not 4000, and it is much more likely lower than that based on the sample.


##Based on the results of the hypothesis test and the decision in step 4, we can draw the following conclusion:


##Conclusion: There is strong evidence to reject the null hypothesis that the true mean weekly operating cost is 4000. Given the extremely low probability of obtaining a sample mean of 3050 or lower if the true mean were indeed 4000, it is highly unlikely that the theoretical mean weekly cost of 4000 is accurate.
